```python
df1 = pd.read_json("US_category_id.json")
```

```python
categories = pd.json_normalize(df1['items'])
```

```python
categories.head()
```

```python
categories = categories[['id','snippet.title']]
```

```python
categories = categories.rename(columns = {'id':'category_id', 'snippet.title':'category_name'})
```

```python
categories['category_id']= categories['category_id'].astype('int64')
```

```python
df_merged = df.merge(categories, on='category_id', how = 'left')
```

```python
len(df)
```

```python
df_merged['category_name'].isna().sum()
```

```python
df_merged.drop_duplicates(subset=['video_id','trending_date'],  keep='last',inplace=True)
```

```python
us_shares = df_merged['category_name'].value_counts(normalize=True)*100
```

```python
us_shares
```

```python
us_shares.sort_values().plot(
    kind='barh',
    title="Share of US trending appearances by category (%)",
    figsize=(6,8),
)
```

```python
unique_df = df_merged.drop_duplicates(subset='video_id',  keep='last')
```

```python
unique_us_shares =unique_df['category_name'].value_counts(normalize = True)*100
```

```python
unique_us_shares
```

```python
unique_us_shares.sort_values().plot(
    kind='barh',
    title="Share of US trending unique appearances by category (%)",
    figsize=(6,8),
)
```

```python
#Question 3: Which category gets the most attention from viewers?
```

```python
max_views = df_merged.groupby('video_id')['views'].max().reset_index()
```

```python
max_views = max_views.reset_index() 
```

```python
views_by_video = max_views.merge(
    unique_df[['video_id', 'category_name']],
    on='video_id',
    how='left'
)
```

```python
views_by_video
```

```python
views_by_video.groupby('category_name')['views'].sum().sort_values(ascending = False)
```

```python
views_sum = views_by_video.groupby("category_name")["views"].sum()
views_share = views_sum / views_sum.sum() * 100
```

```python
views_share.sort_values(ascending = False)
```

```python
views_share.sort_values().plot(
    kind='barh',
    title="Share of US trending unique appearances by views (%)",
    figsize=(6,8),
)
```

```python
#Results 3: 
#Entertainment is now second, but Music % increased to leading position. 
#Music overrun Entertainment’s views with only about half as many videos. So a typical Music video gets far more views
#Comedy and Film & Animation rise. Film & Animation is third by views, and Comedy is fourth, Howto & Style went down to 7th place.
```

```python
views_median = views_by_video.groupby('category_name')['views'].median().sort_values(ascending = False)
```

```python
views_median
```

```python
#Results 4
#Gaming is first by median, but it had a small share of videos and of total views
#Music is second. It was also first by total views, so it’s consistently strong.
#Entertainment went down to 6th.
```

```python
views_mean = views_by_video.groupby('category_name')['views'].mean().sort_values(ascending = False)
```

```python
## Data and method
#Source: Kaggle "Trending YouTube Video Statistics", US file, 14 Nov 2017 to 14 Jun 2018 (40,899 rows after removing 50 duplicates; 6,351 unique videos).
#One row is one video on one day of the trending list.
#Three measures: **appearances** (share of rows), **unique videos** (each video counted once), **views** (each video's highest view count, summed by category).


##**Question:** Which video categories dominate the US trending list, and does the answer depend on how "dominate" is measured?

## Answer
#It depends on the measure. 
#Entertainment is the largest category by appearances (24.3%) and by unique videos (25.5%), 
#but Music is the largest by views (38.7%) with only 12.6% of unique videos.


#**Entertainment:** 24.3% of appearances, 25.5% of videos, 22.7% of views. Its views roughly match its video count.
#**Music:** 12.6% of videos but 38.7% of views. Its median video has about 1.10M views (2nd highest), and its mean is about 5.5 times its median. The top 10 videos hold 24.5% of Music's views [compare: Entertainment top 10 = X%].
#**News & Politics:** 7.9% of videos but only 1.9% of views (median about 179k).
#**Film & Animation:** 5.0% of videos but 6.6% of views (median about 911k, 3rd highest).
#**Comedy:** 8.4%, 8.6%, 6.2%, the only category with the same rank (4th) on all three measures.
```
